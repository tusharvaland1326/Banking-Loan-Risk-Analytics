## 1. Portfolio Overview

The first step is to understand the overall size and composition of the loan portfolio.

## SQL Query

The following query calculates the overall loan portfolio size, number of defaulted applications, number of non-defaulted applications, and overall default rate.

```sql
SELECT
    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS defaulted_applications,

    SUM(
        CASE
            WHEN "TARGET" = '0' THEN 1
            ELSE 0
        END
    ) AS non_defaulted_applications,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS default_rate_percent

FROM staging.application_train;

## Observation

The portfolio contains 307,511 loan applications.

The analysis separates applications into defaulted and non-defaulted groups and calculates the overall default rate.

The default rate provides the baseline risk level for the entire loan portfolio and will be used as a benchmark for the subsequent customer and segment-level risk analysis.

### Key KPI

**Portfolio Default Rate:** 8.07%

**Interpretation:** Approximately 8 out of every 100 applications are associated with loan default in the training dataset.

# 2. Default Rate by Gender

This analysis evaluates whether observed loan default rates differ across customer gender groups.

Instead of comparing only the number of defaults, the analysis calculates the default rate within each gender group.

## SQL Query

```sql
SELECT
    "CODE_GENDER" AS gender,
    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS defaults,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS default_rate_percent

FROM staging.application_train

GROUP BY "CODE_GENDER"

ORDER BY default_rate_percent DESC;
```

## Observation

The analysis compares loan default rates across gender groups.

Default counts are evaluated together with the total number of applications in each group, allowing a fair comparison of observed default risk.

### Business Interpretation

- The gender group with the higher default rate shows relatively higher observed credit risk.
- The result should not be interpreted as a causal relationship.
- Gender is only one characteristic and should be evaluated together with income, credit amount, employment, previous credit history, and other risk factors.

**Status:** ✅ Gender-level default risk analysis completed.

## 17. Age Group Analysis

Age is an important demographic factor that may influence loan repayment behavior. 
This analysis groups applicants into different age categories and compares their 
default rates.

## SQL Query

The following query calculates the total applications, number of defaults, and default rate for each age group.

```sql
SELECT
    CASE
        WHEN "DAYS_BIRTH"::numeric / -365.25 < 25 THEN 'Under 25'
        WHEN "DAYS_BIRTH"::numeric / -365.25 < 35 THEN '25-34'
        WHEN "DAYS_BIRTH"::numeric / -365.25 < 45 THEN '35-44'
        WHEN "DAYS_BIRTH"::numeric / -365.25 < 55 THEN '45-54'
        ELSE '55+'
    END AS age_group,

    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS defaults,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS default_rate_percent

FROM staging.application_train

GROUP BY
    CASE
        WHEN "DAYS_BIRTH"::numeric / -365.25 < 25 THEN 'Under 25'
        WHEN "DAYS_BIRTH"::numeric / -365.25 < 35 THEN '25-34'
        WHEN "DAYS_BIRTH"::numeric / -365.25 < 45 THEN '35-44'
        WHEN "DAYS_BIRTH"::numeric / -365.25 < 55 THEN '45-54'
        ELSE '55+'
    END

ORDER BY default_rate_percent DESC;

## Observation

- **Under 25** applicants have the highest observed default rate at **12.29%**, with **12,233 applications** and **1,504 defaults**.
- **25-34** applicants have a default rate of **10.66%**, based on **72,429 applications** and **7,721 defaults**.
- **35-44** applicants represent the largest age group with **84,261 applications** and have a default rate of **8.41%**.
- **45-54** applicants have a lower default rate of **7.05%**, with **70,190 applications**.
- **55+** applicants have the lowest observed default rate at **5.22%**, with **68,398 applications** and **3,569 defaults**.
- Overall, the results indicate that the observed default rate **decreases as age increases**, with younger applicants showing higher default rates.
- Age group may therefore be a useful factor for further loan-risk analysis, although other financial and demographic factors should also be considered.

# 18. Default Rate by Income Amount

This analysis evaluates the observed loan default rate across different income groups.

The objective is to compare application volume and default rates across different income ranges.



## SQL Query

The following query calculates the total applications, number of defaults, and default rate for each income group.
```sql
SELECT
    CASE
        WHEN "AMT_INCOME_TOTAL"::numeric < 50000 THEN 'Below 50K'
        WHEN "AMT_INCOME_TOTAL"::numeric < 100000 THEN '50K-100K'
        WHEN "AMT_INCOME_TOTAL"::numeric < 200000 THEN '100K-200K'
        WHEN "AMT_INCOME_TOTAL"::numeric < 300000 THEN '200K-300K'
        ELSE '300K+'
    END AS income_group,

    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS defaults,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS default_rate_percent

FROM staging.application_train

GROUP BY
    CASE
        WHEN "AMT_INCOME_TOTAL"::numeric < 50000 THEN 'Below 50K'
        WHEN "AMT_INCOME_TOTAL"::numeric < 100000 THEN '50K-100K'
        WHEN "AMT_INCOME_TOTAL"::numeric < 200000 THEN '100K-200K'
        WHEN "AMT_INCOME_TOTAL"::numeric < 300000 THEN '200K-300K'
        ELSE '300K+'
    END

ORDER BY default_rate_percent DESC;

## Observation

- **100K-200K** income applicants have the highest observed default rate at **8.55%**, with **155,898 applications** and **13,326 defaults**.
- **50K-100K** income applicants have a default rate of **8.25%**, based on **59,181 applications** and **4,882 defaults**.
- **Below 50K** applicants have a default rate of **7.59%**, with **4,517 applications** and **343 defaults**.
- **200K-300K** applicants have a default rate of **7.55%**, with **65,176 applications** and **4,921 defaults**.
- **300K+** income applicants have the lowest observed default rate at **5.95%**, with **22,739 applications** and **1,353 defaults**.
- Overall, the results show that observed default rates vary across income groups, with the **300K+ income group showing the lowest default rate**.
- Income level may therefore be a useful factor for further loan-risk analysis, although other financial and demographic factors should also be considered.

# 19. Default Rate by Credit Amount

This analysis evaluates the observed loan default rate across different credit amount groups.

The objective is to compare application volume and default rates across credit amount ranges.


## SQL Query

The following query calculates the total applications, number of defaults, and default rate for each credit amount group.
    
```sql
                                                                                            SELECT
    CASE
        WHEN "AMT_CREDIT"::numeric < 100000 THEN 'Below 100K'
        WHEN "AMT_CREDIT"::numeric < 200000 THEN '100K-200K'
        WHEN "AMT_CREDIT"::numeric < 300000 THEN '200K-300K'
        WHEN "AMT_CREDIT"::numeric < 500000 THEN '300K-500K'
        ELSE '500K+'
    END AS credit_group,

    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS defaults,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS default_rate_percent

FROM staging.application_train

GROUP BY
    CASE
        WHEN "AMT_CREDIT"::numeric < 100000 THEN 'Below 100K'
        WHEN "AMT_CREDIT"::numeric < 200000 THEN '100K-200K'
        WHEN "AMT_CREDIT"::numeric < 300000 THEN '200K-300K'
        WHEN "AMT_CREDIT"::numeric < 500000 THEN '300K-500K'
        ELSE '500K+'
    END

ORDER BY default_rate_percent DESC;

## Observation

- **300K–500K** credit applications have the highest observed default rate at **9.93%**, with **58,376 applications** and **5,794 defaults**.
- **200K–300K** credit applications have a default rate of **7.88%**, with **54,813 applications**.
- The **500K+** group has the largest application volume with **158,178 applications** and a default rate of **7.73%**.
- **100K–200K** credit applications have a lower default rate of **7.16%**, with **30,140 applications**.
- **Below 100K** has the lowest observed default rate at **5.51%**, with only **6,004 applications**.
- Overall, the results show that default rates vary across credit amount ranges, with the **300K–500K group showing the highest observed risk**.
- Credit amount may therefore be a useful factor for further loan-risk analysis.

# 20. Default Rate by Annuity Amount

This analysis evaluates the observed loan default rate across different annuity amount groups.

The objective is to compare application volume and default rates across different loan annuity ranges.

The analysis groups loan applications based on the amount of the loan annuity and calculates the total applications, number of defaults, and default rate for each group.

## SQL Query

The following SQL query groups applications into different annuity amount ranges and calculates the total number of applications, number of defaults, and default rate for each group.

```sql
SELECT
    CASE
        WHEN "AMT_ANNUITY"::numeric < 20000 THEN 'Below 20K'
        WHEN "AMT_ANNUITY"::numeric < 40000 THEN '20K-40K'
        WHEN "AMT_ANNUITY"::numeric < 60000 THEN '40K-60K'
        WHEN "AMT_ANNUITY"::numeric < 100000 THEN '60K-100K'
        ELSE '100K+'
    END AS annuity_group,

    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS defaults,

    ROUND(
        100.0 * SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS default_rate_percent

FROM staging.application_train

GROUP BY
    CASE
        WHEN "AMT_ANNUITY"::numeric < 20000 THEN 'Below 20K'
        WHEN "AMT_ANNUITY"::numeric < 40000 THEN '20K-40K'
        WHEN "AMT_ANNUITY"::numeric < 60000 THEN '40K-60K'
        WHEN "AMT_ANNUITY"::numeric < 100000 THEN '60K-100K'
        ELSE '100K+'
    END

ORDER BY default_rate_percent DESC;

### Observation

- **20K–40K annuity amounts** have the highest default rate at **8.94%**, with **151,028 applications** and **13,509 defaults**.
- **Below 20K** represents the second-largest group with **106,505 applications** and has a default rate of **7.59%**.
- **40K–60K** has **41,834 applications** and a default rate of **6.92%**.
- **60K–100K** has a lower default rate of **4.26%**, based on **7,627 applications**.
- **100K+** has the lowest observed default rate at **1.93%**, with only **517 applications** and **10 defaults**, so this result should be interpreted cautiously.
- Overall, the observed default rate **decreases as the annuity amount increases**, with the exception that the 20K–40K group has the highest default rate.
- Annuity amount appears to be a useful factor for further loan-risk analysis, although groups with small sample sizes should be interpreted carefully.

# 21. Default Rate by Goods Price

This analysis evaluates the observed loan default rate across different goods price groups.

The objective is to compare application volume and default rates across different goods price ranges.

The analysis groups loan applications based on the price of the goods associated with the loan and calculates the total applications, number of defaults, and default rate for each group.

## SQL Query

The following SQL query groups applications into different goods price ranges and calculates the total number of applications, number of defaults, and default rate for each group.

```sql
SELECT
    CASE
        WHEN "AMT_GOODS_PRICE"::numeric < 100000 THEN 'Below 100K'
        WHEN "AMT_GOODS_PRICE"::numeric < 200000 THEN '100K-200K'
        WHEN "AMT_GOODS_PRICE"::numeric < 300000 THEN '200K-300K'
        WHEN "AMT_GOODS_PRICE"::numeric < 500000 THEN '300K-500K'
        ELSE '500K+'
    END AS goods_price_group,

    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS defaults,

    ROUND(
        100.0 * SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS default_rate_percent

FROM staging.application_train

GROUP BY
    CASE
        WHEN "AMT_GOODS_PRICE"::numeric < 100000 THEN 'Below 100K'
        WHEN "AMT_GOODS_PRICE"::numeric < 200000 THEN '100K-200K'
        WHEN "AMT_GOODS_PRICE"::numeric < 300000 THEN '200K-300K'
        WHEN "AMT_GOODS_PRICE"::numeric < 500000 THEN '300K-500K'
        ELSE '500K+'
    END

ORDER BY default_rate_percent DESC;

### Observation

- **300K–500K goods prices** have the highest observed default rate at **10.40%**, with **78,192 applications** and **8,132 defaults**.
- **200K–300K** has a default rate of **8.42%**, based on **62,761 applications** and **5,283 defaults**.
- **100K–200K** has a default rate of **7.69%**, with **32,956 applications** and **2,534 defaults**.
- **500K+** represents the largest group with **124,893 applications** and has a lower default rate of **6.69%**.
- **Below 100K** has the lowest observed default rate at **6.04%**, with **8,709 applications** and **526 defaults**.
- Overall, default rates are highest in the **300K–500K** range and generally lower at the higher and lower ends of the goods price range.
- Goods price appears to be a useful factor for further loan-risk analysis, although default rates should be considered together with application volume.

# 22. Default Rate by Credit Income Ratio

This analysis evaluates the observed loan default rate across different credit-to-income ratio groups.

The objective is to compare application volume and default rates across different credit-to-income ranges.

The analysis groups loan applications based on the relationship between the credit amount and the applicant's income, then calculates the total applications, number of defaults, and default rate for each group.

## SQL Query

The following SQL query groups applications based on the credit-to-income ratio and calculates the total number of applications, number of defaults, and default rate for each group.

```sql
SELECT
    CASE
        WHEN ("AMT_CREDIT"::numeric / NULLIF("AMT_INCOME_TOTAL"::numeric, 0)) < 1
            THEN 'Below 1'
        WHEN ("AMT_CREDIT"::numeric / NULLIF("AMT_INCOME_TOTAL"::numeric, 0)) < 2
            THEN '1-2'
        WHEN ("AMT_CREDIT"::numeric / NULLIF("AMT_INCOME_TOTAL"::numeric, 0)) < 3
            THEN '2-3'
        WHEN ("AMT_CREDIT"::numeric / NULLIF("AMT_INCOME_TOTAL"::numeric, 0)) < 4
            THEN '3-4'
        ELSE '4+'
    END AS credit_income_ratio_group,

    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS defaults,

    ROUND(
        100.0 * SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS default_rate_percent

FROM staging.application_train

GROUP BY
    CASE
        WHEN ("AMT_CREDIT"::numeric / NULLIF("AMT_INCOME_TOTAL"::numeric, 0)) < 1
            THEN 'Below 1'
        WHEN ("AMT_CREDIT"::numeric / NULLIF("AMT_INCOME_TOTAL"::numeric, 0)) < 2
            THEN '1-2'
        WHEN ("AMT_CREDIT"::numeric / NULLIF("AMT_INCOME_TOTAL"::numeric, 0)) < 3
            THEN '2-3'
        WHEN ("AMT_CREDIT"::numeric / NULLIF("AMT_INCOME_TOTAL"::numeric, 0)) < 4
            THEN '3-4'
        ELSE '4+'
    END

ORDER BY default_rate_percent DESC;

### Observation

- **3–4 credit-to-income ratio** has the highest observed default rate at **8.85%**, with **50,380 applications** and **4,461 defaults**.
- **2–3** has a similar default rate of **8.70%**, based on **66,240 applications** and **5,762 defaults**.
- **4+** represents the largest group with **119,855 applications** and has a default rate of **7.75%**.
- **1–2** has a default rate of **7.72%**, with **56,798 applications** and **4,384 defaults**.
- **Below 1** has the lowest observed default rate at **6.55%**, based on **14,238 applications** and **933 defaults**.
- Overall, the results show that **higher credit relative to income is generally associated with higher observed default risk**, particularly in the **2–4 ratio range**.
- Credit-to-income ratio appears to be a useful factor for further loan-risk analysis.

# 23. Default Rate by Annuity-to-Credit Ratio

This analysis evaluates the observed loan default rate across different annuity-to-credit ratio groups.

The objective is to compare application volume and default rates across different annuity-to-credit ratio ranges.

The analysis groups loan applications based on the relationship between the annuity amount and the credit amount, then calculates the total applications, number of defaults, and default rate for each group.

## SQL Query

The following SQL query groups applications based on the annuity-to-credit ratio and calculates the total number of applications, number of defaults, and default rate for each group.

```sql
SELECT
    CASE
        WHEN ("AMT_ANNUITY"::numeric / NULLIF("AMT_CREDIT"::numeric, 0)) < 0.10
            THEN 'Below 10%'
        WHEN ("AMT_ANNUITY"::numeric / NULLIF("AMT_CREDIT"::numeric, 0)) < 0.20
            THEN '10%-20%'
        WHEN ("AMT_ANNUITY"::numeric / NULLIF("AMT_CREDIT"::numeric, 0)) < 0.30
            THEN '20%-30%'
        WHEN ("AMT_ANNUITY"::numeric / NULLIF("AMT_CREDIT"::numeric, 0)) < 0.40
            THEN '30%-40%'
        ELSE '40%+'
    END AS annuity_credit_ratio_group,

    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS defaults,

    ROUND(
        100.0 * SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS default_rate_percent

FROM staging.application_train

GROUP BY
    CASE
        WHEN ("AMT_ANNUITY"::numeric / NULLIF("AMT_CREDIT"::numeric, 0)) < 0.10
            THEN 'Below 10%'
        WHEN ("AMT_ANNUITY"::numeric / NULLIF("AMT_CREDIT"::numeric, 0)) < 0.20
            THEN '10%-20%'
        WHEN ("AMT_ANNUITY"::numeric / NULLIF("AMT_CREDIT"::numeric, 0)) < 0.30
            THEN '20%-30%'
        WHEN ("AMT_ANNUITY"::numeric / NULLIF("AMT_CREDIT"::numeric, 0)) < 0.40
            THEN '30%-40%'
        ELSE '40%+'
    END

ORDER BY default_rate_percent DESC;

### Observation

- **Below 10%** has the highest observed default rate at **8.20%**, with **287,832 applications** and **23,609 defaults**.
- **10%–20%** has a lower default rate of **6.18%**, based on **19,667 applications** and **1,216 defaults**.
- The **40%+** group contains only **12 applications** and **0 defaults**, resulting in a **0.00%** default rate. This result should not be used for strong conclusions because of the very small sample size.
- The available results show that applications with an **annuity-to-credit ratio below 10% have a higher observed default rate** than the 10%–20% group.
- Overall, annuity-to-credit ratio may be a useful factor for further loan-risk analysis, but very small groups should be interpreted cautiously.

# 24. Default Rate by Credit-to-Goods Price Ratio

This analysis evaluates the observed loan default rate across different credit-to-goods price ratio groups.

The objective is to compare application volume and default rates across different credit-to-goods price ratio ranges.

The analysis groups loan applications based on the relationship between the credit amount and the goods price, then calculates the total applications, number of defaults, and default rate for each group.

## SQL Query

The following SQL query groups applications based on the credit-to-goods price ratio and calculates the total number of applications, number of defaults, and default rate for each group.

```sql
SELECT
    CASE
        WHEN ("AMT_CREDIT"::numeric / NULLIF("AMT_GOODS_PRICE"::numeric, 0)) < 0.75
            THEN 'Below 75%'
        WHEN ("AMT_CREDIT"::numeric / NULLIF("AMT_GOODS_PRICE"::numeric, 0)) < 1.00
            THEN '75%-100%'
        WHEN ("AMT_CREDIT"::numeric / NULLIF("AMT_GOODS_PRICE"::numeric, 0)) < 1.25
            THEN '100%-125%'
        WHEN ("AMT_CREDIT"::numeric / NULLIF("AMT_GOODS_PRICE"::numeric, 0)) < 1.50
            THEN '125%-150%'
        ELSE '150%+'
    END AS credit_goods_ratio_group,

    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS defaults,

    ROUND(
        100.0 * SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS default_rate_percent

FROM staging.application_train

GROUP BY
    CASE
        WHEN ("AMT_CREDIT"::numeric / NULLIF("AMT_GOODS_PRICE"::numeric, 0)) < 0.75
            THEN 'Below 75%'
        WHEN ("AMT_CREDIT"::numeric / NULLIF("AMT_GOODS_PRICE"::numeric, 0)) < 1.00
            THEN '75%-100%'
        WHEN ("AMT_CREDIT"::numeric / NULLIF("AMT_GOODS_PRICE"::numeric, 0)) < 1.25
            THEN '100%-125%'
        WHEN ("AMT_CREDIT"::numeric / NULLIF("AMT_GOODS_PRICE"::numeric, 0)) < 1.50
            THEN '125%-150%'
        ELSE '150%+'
    END

ORDER BY default_rate_percent DESC;

### Observation

- The **150%+** credit-to-goods price ratio group has the highest default rate at **12.93%**, based on **2,853 applications**.
- The **125%-150%** group also has a high default rate of **12.10%**, with **39,107 applications** and **4,732 defaults**.
- The **100%-125%** group represents the largest volume with **265,291 applications** and has a default rate of **7.43%**.
- The **75%-100%** group has a default rate of **8.54%**, but the sample contains only **82 applications**, so the result should be interpreted cautiously.
- The **Below 75%** group has the lowest observed default rate at **3.37%**, based on **178 applications**.
- Overall, higher credit-to-goods price ratios, particularly above **125%**, are associated with higher observed default rates.
- Small groups such as **75%-100%** and **Below 75%** should be interpreted carefully because of their limited sample sizes.

# 25. Default Rate by Credit-to-Income Ratio

This analysis evaluates the observed loan default rate across different credit-to-income ratio groups.

The objective is to compare application volume and default rates across different levels of credit relative to applicant income.

## SQL Query

The following query groups applicants based on their credit-to-income ratio and calculates the total applications, number of defaults, and default rate for each group.

```sql
SELECT
    CASE
        WHEN "AMT_CREDIT"::numeric / NULLIF("AMT_INCOME_TOTAL"::numeric, 0) < 1
            THEN 'Below 1'
        WHEN "AMT_CREDIT"::numeric / NULLIF("AMT_INCOME_TOTAL"::numeric, 0) < 2
            THEN '1-2'
        WHEN "AMT_CREDIT"::numeric / NULLIF("AMT_INCOME_TOTAL"::numeric, 0) < 3
            THEN '2-3'
        WHEN "AMT_CREDIT"::numeric / NULLIF("AMT_INCOME_TOTAL"::numeric, 0) < 4
            THEN '3-4'
        ELSE '4+'
    END AS credit_income_ratio_group,

    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS defaults,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS default_rate_percent

FROM staging.application_train

GROUP BY
    CASE
        WHEN "AMT_CREDIT"::numeric / NULLIF("AMT_INCOME_TOTAL"::numeric, 0) < 1
            THEN 'Below 1'
        WHEN "AMT_CREDIT"::numeric / NULLIF("AMT_INCOME_TOTAL"::numeric, 0) < 2
            THEN '1-2'
        WHEN "AMT_CREDIT"::numeric / NULLIF("AMT_INCOME_TOTAL"::numeric, 0) < 3
            THEN '2-3'
        WHEN "AMT_CREDIT"::numeric / NULLIF("AMT_INCOME_TOTAL"::numeric, 0) < 4
            THEN '3-4'
        ELSE '4+'
    END

ORDER BY default_rate_percent DESC;

### Observation

- The **3-4** credit-to-income ratio group has the highest observed default rate at **8.85%**, with **50,380 applications** and **4,461 defaults**.
- The **2-3** group has a similar default rate of **8.70%**, based on **66,240 applications**.
- The **4+** group represents the largest volume with **119,855 applications** and has a default rate of **7.75%**.
- The **1-2** group has a default rate of **7.72%**, with **56,798 applications**.
- The **Below 1** group has the lowest observed default rate at **6.55%**, based on **14,238 applications**.
- Overall, applicants with higher credit relative to income generally show higher observed default rates, with the **3-4** and **2-3** groups having the highest rates.
- The differences are moderate, so credit-to-income ratio should be considered alongside other loan-risk factors.

# 26. Default Rate by Age Group

This analysis evaluates the observed loan default rate across different age groups.

The objective is to compare application volume and default rates across different applicant age ranges and identify whether age appears to be associated with loan default risk.

## SQL Query

The following query groups applicants into age ranges and calculates the total applications, number of defaults, and default rate for each age group.

```sql
SELECT
    CASE
        WHEN "DAYS_BIRTH"::numeric / -365.25 < 25 THEN 'Under 25'
        WHEN "DAYS_BIRTH"::numeric / -365.25 < 35 THEN '25-34'
        WHEN "DAYS_BIRTH"::numeric / -365.25 < 45 THEN '35-44'
        WHEN "DAYS_BIRTH"::numeric / -365.25 < 55 THEN '45-54'
        ELSE '55+'
    END AS age_group,

    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS defaults,

    ROUND(
        100.0 * SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS default_rate_percent

FROM staging.application_train

GROUP BY
    CASE
        WHEN "DAYS_BIRTH"::numeric / -365.25 < 25 THEN 'Under 25'
        WHEN "DAYS_BIRTH"::numeric / -365.25 < 35 THEN '25-34'
        WHEN "DAYS_BIRTH"::numeric / -365.25 < 45 THEN '35-44'
        WHEN "DAYS_BIRTH"::numeric / -365.25 < 55 THEN '45-54'
        ELSE '55+'
    END

ORDER BY default_rate_percent DESC;

## Observation

- **Under 25** applicants have the highest observed default rate at **12.29%**, with **12,233 applications** and **1,504 defaults**.
- Applicants aged **25-34** have a default rate of **10.66%**, based on **72,429 applications**.
- The **35-44** age group represents the largest group with **84,261 applications** and has a default rate of **8.41%**.
- Applicants aged **45-54** have a lower default rate of **7.05%**, with **70,190 applications**.
- The **55+** group has the lowest observed default rate at **5.22%**, based on **68,398 applications**.
- Overall, the observed default rate decreases as applicant age increases, with younger applicants showing higher default rates.
- Age appears to be a potentially useful factor for further loan-risk analysis.

# 27. Default Rate by Income Group

This analysis evaluates the observed loan default rate across different income groups.

The objective is to compare application volume and default rates across different applicant income ranges and identify whether income level appears to be associated with loan default risk.

## SQL Query

The following query groups applicants based on their income level and calculates the total applications, number of defaults, and default rate for each income group.

```sql
SELECT
    CASE
        WHEN "AMT_INCOME_TOTAL"::numeric < 50000 THEN 'Below 50K'
        WHEN "AMT_INCOME_TOTAL"::numeric < 100000 THEN '50K-100K'
        WHEN "AMT_INCOME_TOTAL"::numeric < 200000 THEN '100K-200K'
        WHEN "AMT_INCOME_TOTAL"::numeric < 300000 THEN '200K-300K'
        ELSE '300K+'
    END AS income_group,

    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS defaults,

    ROUND(
        100.0 * SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS default_rate_percent

FROM staging.application_train

GROUP BY
    CASE
        WHEN "AMT_INCOME_TOTAL"::numeric < 50000 THEN 'Below 50K'
        WHEN "AMT_INCOME_TOTAL"::numeric < 100000 THEN '50K-100K'
        WHEN "AMT_INCOME_TOTAL"::numeric < 200000 THEN '100K-200K'
        WHEN "AMT_INCOME_TOTAL"::numeric < 300000 THEN '200K-300K'
        ELSE '300K+'
    END

ORDER BY default_rate_percent DESC;

## Observation

- **100K-200K** applicants have the highest observed default rate at **8.55%**, with **155,898 applications** and **13,326 defaults**.
- The **50K-100K** income group has a default rate of **8.25%**, based on **59,181 applications**.
- Applicants with income **Below 50K** have a default rate of **7.59%**, with **4,517 applications**.
- The **200K-300K** group has a default rate of **7.55%**, based on **65,176 applications**.
- The **300K+** income group has the lowest observed default rate at **5.95%**, with **22,739 applications** and **1,353 defaults**.
- Overall, higher-income applicants, particularly those earning **300K+**, show a lower observed default rate.
- Income level may therefore be a useful factor for further loan-risk analysis.

# 28. Default Rate by Credit Amount

This analysis evaluates the observed loan default rate across different credit amount groups.

The objective is to compare application volume and default rates across different loan credit ranges and identify whether the credit amount appears to be associated with loan default risk.

## SQL Query

The following query groups applicants based on their credit amount and calculates the total applications, number of defaults, and default rate for each credit amount group.

```sql
SELECT
    CASE
        WHEN "AMT_CREDIT"::numeric < 100000 THEN 'Below 100K'
        WHEN "AMT_CREDIT"::numeric < 200000 THEN '100K-200K'
        WHEN "AMT_CREDIT"::numeric < 300000 THEN '200K-300K'
        WHEN "AMT_CREDIT"::numeric < 500000 THEN '300K-500K'
        ELSE '500K+'
    END AS credit_group,

    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS defaults,

    ROUND(
        100.0 * SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS default_rate_percent

FROM staging.application_train

GROUP BY
    CASE
        WHEN "AMT_CREDIT"::numeric < 100000 THEN 'Below 100K'
        WHEN "AMT_CREDIT"::numeric < 200000 THEN '100K-200K'
        WHEN "AMT_CREDIT"::numeric < 300000 THEN '200K-300K'
        WHEN "AMT_CREDIT"::numeric < 500000 THEN '300K-500K'
        ELSE '500K+'
    END

ORDER BY default_rate_percent DESC;

## Observation

- **300K-500K** applicants have the highest observed default rate at **9.93%**, with **58,376 applications** and **5,794 defaults**.
- The **200K-300K** credit group has a default rate of **7.88%**, based on **54,813 applications**.
- The **500K+** group represents the largest volume with **158,178 applications** and has a default rate of **7.73%**.
- The **100K-200K** group has a default rate of **7.16%**, with **30,140 applications**.
- The **Below 100K** group has the lowest observed default rate at **5.51%**, based on **6,004 applications**.
- Overall, the **300K-500K** credit group shows the highest observed default rate, while the **Below 100K** group shows the lowest.
- Credit amount appears to be a potentially useful factor for further loan-risk analysis, although the relationship is not strictly linear across all groups.

# 29. Default Rate by Annuity / Payment Amount

This analysis evaluates the observed loan default rate across different annuity or payment amount groups.

The objective is to compare application volume and default rates across different payment amount ranges and identify whether the annuity amount appears to be associated with loan default risk.

## SQL Query

The following query groups applicants based on their annuity or payment amount and calculates the total applications, number of defaults, and default rate for each annuity group.

```sql
SELECT
    CASE
        WHEN "AMT_ANNUITY"::numeric < 20000 THEN 'Below 20K'
        WHEN "AMT_ANNUITY"::numeric < 40000 THEN '20K-40K'
        WHEN "AMT_ANNUITY"::numeric < 60000 THEN '40K-60K'
        WHEN "AMT_ANNUITY"::numeric < 100000 THEN '60K-100K'
        ELSE '100K+'
    END AS annuity_group,

    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS defaults,

    ROUND(
        100.0 * SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS default_rate_percent

FROM staging.application_train

GROUP BY
    CASE
        WHEN "AMT_ANNUITY"::numeric < 20000 THEN 'Below 20K'
        WHEN "AMT_ANNUITY"::numeric < 40000 THEN '20K-40K'
        WHEN "AMT_ANNUITY"::numeric < 60000 THEN '40K-60K'
        WHEN "AMT_ANNUITY"::numeric < 100000 THEN '60K-100K'
        ELSE '100K+'
    END

ORDER BY default_rate_percent DESC;

## Observation

- The **20K-40K** annuity group has the highest observed default rate at **8.94%**, with **151,028 applications** and **13,509 defaults**.
- The **Below 20K** group has a default rate of **7.59%**, based on **106,505 applications**.
- The **40K-60K** group has a default rate of **6.92%**, with **41,834 applications** and **2,897 defaults**.
- The **60K-100K** group has a lower default rate of **4.26%**, based on **7,627 applications**.
- The **100K+** group has the lowest observed default rate at **1.93%**, but it contains only **517 applications**, so the result should be interpreted cautiously.
- Overall, the observed default rate decreases as the annuity amount increases.
- Annuity or payment amount may therefore be a useful factor for further loan-risk analysis, although small categories should be interpreted carefully.

# 30. Key Risk-Factor Summary

This analysis brings together the main risk factors identified in the previous SQL analyses.

The objective is to summarize the groups with relatively high observed default rates across age, income, credit amount, and annuity amount while considering application volume.

Small groups with very few applications are excluded to avoid drawing conclusions from unstable sample sizes.

## SQL Query

The following query calculates the total applications, number of defaults, and default rate for each annuity amount group.

```sql
SELECT
    CASE
        WHEN "AMT_ANNUITY"::numeric < 20000 THEN 'Below 20K'
        WHEN "AMT_ANNUITY"::numeric < 40000 THEN '20K-40K'
        WHEN "AMT_ANNUITY"::numeric < 60000 THEN '40K-60K'
        WHEN "AMT_ANNUITY"::numeric < 100000 THEN '60K-100K'
        ELSE '100K+'
    END AS annuity_group,

    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS defaults,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS default_rate_percent

FROM staging.application_train

GROUP BY
    CASE
        WHEN "AMT_ANNUITY"::numeric < 20000 THEN 'Below 20K'
        WHEN "AMT_ANNUITY"::numeric < 40000 THEN '20K-40K'
        WHEN "AMT_ANNUITY"::numeric < 60000 THEN '40K-60K'
        WHEN "AMT_ANNUITY"::numeric < 100000 THEN '60K-100K'
        ELSE '100K+'
    END

ORDER BY default_rate_percent DESC;

## Observation

- **20K-40K annuity amounts** have the highest observed default rate at **8.94%**, with **151,028 applications** and **13,509 defaults**.
- **Below 20K** represents the second-largest group with **106,505 applications** and a default rate of **7.59%**.
- **40K-60K** has a default rate of **6.92%**, based on **41,834 applications**.
- **60K-100K** has a lower default rate of **4.26%**, with **7,627 applications**.
- **100K+** has the lowest observed default rate at **1.93%**, but the category contains only **517 applications**, so the result should be interpreted cautiously.
- Overall, the results indicate that **default rates decrease as annuity amounts increase**, although the smallest category should not be used for strong conclusions.

# 31. Default Rate by Employment Duration

This analysis evaluates the observed loan default rate across different employment duration groups.

The objective is to compare application volume and default rates based on how long applicants have been employed.

## SQL Query

The following query calculates the total applications, number of defaults, and default rate for each employment duration group.

```sql
SELECT
    CASE
        WHEN "DAYS_EMPLOYED"::numeric < 2 * 365 THEN 'Below 2 Years'
        WHEN "DAYS_EMPLOYED"::numeric < 5 * 365 THEN '2-5 Years'
        WHEN "DAYS_EMPLOYED"::numeric < 10 * 365 THEN '5-10 Years'
        WHEN "DAYS_EMPLOYED"::numeric < 20 * 365 THEN '10-20 Years'
        ELSE '20+ Years'
    END AS employment_group,

    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS defaults,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS default_rate_percent

FROM staging.application_train

GROUP BY
    CASE
        WHEN "DAYS_EMPLOYED"::numeric < 2 * 365 THEN 'Below 2 Years'
        WHEN "DAYS_EMPLOYED"::numeric < 5 * 365 THEN '2-5 Years'
        WHEN "DAYS_EMPLOYED"::numeric < 10 * 365 THEN '5-10 Years'
        WHEN "DAYS_EMPLOYED"::numeric < 20 * 365 THEN '10-20 Years'
        ELSE '20+ Years'
    END

ORDER BY default_rate_percent DESC;

## Observation

- **Below 2 Years** has the higher observed default rate at **8.66%**, with **252,137 applications** and **21,835 defaults**.
- **20+ Years** has a lower default rate of **5.40%**, with **55,374 applications** and **2,990 defaults**.
- The majority of applications are from applicants with **less than 2 years of employment**, indicating a much larger application volume in this group.
- Overall, the results show that applicants with **shorter employment duration have a higher observed default rate** than applicants with 20+ years of employment.
- This suggests that **employment duration may be a useful factor for further loan-risk analysis**.

# 32. Default Rate by Loan Term

This analysis evaluates the observed loan default rate across different estimated loan-term groups.

The objective is to compare application volume and default rates based on the estimated repayment duration of loans.

## SQL Query

The following query estimates the loan term using the credit amount and annuity amount, then calculates the total applications, number of defaults, and default rate for each loan-term group.

```sql
SELECT
    CASE
        WHEN "AMT_CREDIT"::numeric / NULLIF("AMT_ANNUITY"::numeric, 0) < 24 THEN 'Below 2 Years'
        WHEN "AMT_CREDIT"::numeric / NULLIF("AMT_ANNUITY"::numeric, 0) < 48 THEN '2-4 Years'
        WHEN "AMT_CREDIT"::numeric / NULLIF("AMT_ANNUITY"::numeric, 0) < 72 THEN '4-6 Years'
        WHEN "AMT_CREDIT"::numeric / NULLIF("AMT_ANNUITY"::numeric, 0) < 120 THEN '6-10 Years'
        ELSE '10+ Years'
    END AS loan_term_group,

    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS defaults,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS default_rate_percent

FROM staging.application_train

WHERE "AMT_ANNUITY"::numeric > 0

GROUP BY
    CASE
        WHEN "AMT_CREDIT"::numeric / NULLIF("AMT_ANNUITY"::numeric, 0) < 24 THEN 'Below 2 Years'
        WHEN "AMT_CREDIT"::numeric / NULLIF("AMT_ANNUITY"::numeric, 0) < 48 THEN '2-4 Years'
        WHEN "AMT_CREDIT"::numeric / NULLIF("AMT_ANNUITY"::numeric, 0) < 72 THEN '4-6 Years'
        WHEN "AMT_CREDIT"::numeric / NULLIF("AMT_ANNUITY"::numeric, 0) < 120 THEN '6-10 Years'
        ELSE '10+ Years'
    END

ORDER BY default_rate_percent DESC;

## Observation

- **Below 2 Years** has the higher observed default rate at **8.81%**, with **206,913 applications** and **18,222 defaults**.
- **2-4 Years** has a lower default rate of **6.56%**, with **100,586 applications** and **6,603 defaults**.
- The majority of applications fall into the **Below 2 Years** loan-term group.
- Overall, the results indicate that **shorter estimated loan terms are associated with a higher observed default rate** in this dataset.
- Loan term may therefore be a useful factor for further **loan-risk analysis**, although this grouping is based on an estimated term calculated from credit amount and annuity amount.

# 33. Key Risk-Factor Summary

This analysis summarizes the main risk factors identified from the previous SQL analyses.

The objective is to highlight the categories with relatively higher observed default rates and identify the key factors that may be useful for further loan-risk assessment.
    

## SQL Query

The following query calculates the overall number of applications, total defaults, and overall default rate for the loan portfolio.

```sql
SELECT
    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS total_defaults,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS overall_default_rate_percent

FROM staging.application_train;

## Observation

- The dataset contains **307,511 loan applications** in total.
- There are **24,825 defaulted applications**.
- The overall observed default rate is **8.07%**.
- This overall rate provides a useful baseline for comparing the default rates identified across different customer and loan characteristics.
- The previous analyses show that factors such as **age, income, credit amount, annuity amount, employment duration, education, housing, occupation, and loan term** can have different observed default rates.
- Overall, these findings can be used to identify **higher-risk customer segments** and support further loan-risk analysis.

# 34. Final SQL Insights

This section summarizes the key findings identified throughout the SQL analysis.

The objective is to consolidate the major default-rate patterns and highlight the most important factors associated with loan default risk.

## SQL Query

The following query provides the overall loan portfolio performance, including total applications, total defaults, and the overall default rate.

```sql
SELECT
    COUNT(*) AS total_applications,

    SUM(
        CASE
            WHEN "TARGET" = '1' THEN 1
            ELSE 0
        END
    ) AS total_defaults,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN "TARGET" = '1' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS overall_default_rate_percent

FROM staging.application_train;

## Final SQL Insights

- The dataset contains **307,511 loan applications** and **24,825 defaults**, resulting in an overall default rate of **8.07%**.
- **Age analysis** shows that younger applicants, particularly those under 25, have a relatively higher observed default rate.
- **Income analysis** shows differences in default rates across income groups, with the 100K-200K group showing the highest rate among the analyzed income categories.
- **Credit amount analysis** indicates that the 300K-500K credit group has the highest observed default rate at **9.93%**.
- **Annuity analysis** shows the highest default rate in the 20K-40K group at **8.94%**, while the 100K+ group has the lowest rate at **1.93%**.
- **Employment duration analysis** shows a higher default rate among applicants with shorter employment duration.
- **Loan-term analysis** shows a higher observed default rate for the Below 2 Years group at **8.81%**.
- Other categorical analyses, including **education, housing, occupation, family status, and gender**, also show differences in observed default rates between groups.
- Overall, the analysis indicates that **customer characteristics and loan characteristics can be useful for identifying segments with different levels of observed default risk**.
- These findings can be used as a foundation for the **Power BI dashboard and final project recommendations**.

# 35. Power BI Dashboard

The Power BI dashboard will visualize the key findings from the loan application analysis.

The objective is to present application volume, default counts, default rates, and major risk factors in an interactive and easy-to-understand dashboard.
