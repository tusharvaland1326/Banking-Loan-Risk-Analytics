## Why are all columns created as TEXT?

During the staging process, every column is initially imported as TEXT.

This prevents import failures caused by inconsistent data types.

The correct data types will be assigned later during the SQL Data Cleaning and Data Warehouse phases.

````python
import pandas as pd

# Read only the header
df = pd.read_csv("../DATA/application_train.csv", nrows=1)

print("Total Columns:", len(df.columns))
````

## Observation

The CREATE TABLE statement is generated automatically using Python.

This approach eliminates manual errors and ensures that all dataset columns are imported accurately into PostgreSQL.

````python
sql = "CREATE TABLE staging.application_train (\n"

for col in df.columns:
    sql += f'    "{col}" TEXT,\n'

sql = sql[:-2] + "\n);"

print(sql)
````

# Duplicate Record Validation

## Dataset Used

Schema:

- staging

Table:

- application_train

---

## SQL Query

```sql
SELECT
    "SK_ID_CURR",
    COUNT(*) AS duplicate_count
FROM staging.application_train
GROUP BY "SK_ID_CURR"
HAVING COUNT(*) > 1;
```

## Observation

- Primary Key Checked: **SK_ID_CURR**
- Duplicate Records Found: **0**
- Status: ✅ No duplicate customer IDs found.

The dataset contains unique customer records and is ready for further SQL analysis.
