# Data Quality Checks

## Objective

The purpose of this phase is to identify invalid, inconsistent, and unrealistic values within the dataset before performing analytical or predictive tasks.

The following business rules will be validated:

- Target values
- Gender values
- Contract types
- Income values
- Credit values
- Goods price values
- Number of children
- Family members

These checks help ensure data reliability for downstream SQL analysis, machine learning, and Power BI reporting.

## Validation 1: Target Column

### SQL Query

```sql
SELECT
    "TARGET",
    COUNT(*) AS total_customers
FROM staging.application_train
GROUP BY "TARGET"
ORDER BY "TARGET";
```

### Observation

- TARGET should contain only values 0 and 1.
- 0 represents customers who successfully repaid their loans.
- 1 represents customers who defaulted on their loans.
- Status: ✅ Valid target values confirmed.

## Validation 2: Gender Values

### SQL Query

```sql
SELECT
    "CODE_GENDER",
    COUNT(*) AS total_customers
FROM staging.application_train
GROUP BY "CODE_GENDER"
ORDER BY total_customers DESC;
```

### Observation

- Expected gender values include M and F.
- Additional values, if present, should be reviewed for data quality issues.
- Status: ✅ Gender values validated.

## Validation 3: Loan Contract Types

### SQL Query

```sql
SELECT
    "NAME_CONTRACT_TYPE",
    COUNT(*) AS total_loans
FROM staging.application_train
GROUP BY "NAME_CONTRACT_TYPE";
```

### Observation

- Loan contract types are grouped into available categories.
- This confirms there are no unexpected contract labels.
- Status: ✅ Contract types validated.

## Validation 4: Negative Income Check

### SQL Query

```sql
SELECT
    COUNT(*) AS negative_income_records
FROM staging.application_train
WHERE CAST("AMT_INCOME_TOTAL" AS NUMERIC) < 0;
```

### Observation

- Annual income should never be negative.
- Any negative value indicates invalid or corrupted data.
- Status: ✅ Income values validated.

## Validation 5: Credit Amount Validation

### SQL Query

```sql
SELECT
    COUNT(*) AS invalid_credit_amount
FROM staging.application_train
WHERE CAST("AMT_CREDIT" AS NUMERIC) <= 0;
```

### Observation

- Loan amount must always be greater than zero.
- Zero or negative credit values indicate invalid records.
- Status: ✅ Credit amount validated.

## Validation 6: Goods Price Validation

### SQL Query

```sql
SELECT
    COUNT(*) AS invalid_goods_price
FROM staging.application_train
WHERE CAST("AMT_GOODS_PRICE" AS NUMERIC) <= 0;
```

### Observation

- Goods price should always be greater than zero.
- Invalid values may affect loan-to-value calculations.
- Status: ✅ Goods price validated.

## Validation 7: Number of Children

### SQL Query

```sql
SELECT
    MAX(CAST("CNT_CHILDREN" AS NUMERIC)) AS maximum_children,
    MIN(CAST("CNT_CHILDREN" AS NUMERIC)) AS minimum_children
FROM staging.application_train;
```

### Observation

- Checked the range of children values.
- Extremely high values should be reviewed for accuracy.
- Status: ✅ Children count validated.

## Validation 8: Family Members

### SQL Query

```sql
SELECT
    MAX(CAST("CNT_FAM_MEMBERS" AS NUMERIC)) AS maximum_family_members,
    MIN(CAST("CNT_FAM_MEMBERS" AS NUMERIC)) AS minimum_family_members
FROM staging.application_train;
```

### Observation

- Verified the range of family member values.
- Extreme values should be investigated before analysis.
- Status: ✅ Family member values validated.
