## 10. Target Variable Analysis

### Objective

Analyze the loan repayment target variable.

```python
df["TARGET"].value_counts()
```

```python
df["TARGET"].value_counts(normalize=True) * 100
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x="TARGET", data=df)

plt.title("Loan Repayment Status")
plt.xlabel("Target")
plt.ylabel("Count")

plt.show()
```

### Observation

- Target = 0 → Customer repaid the loan successfully.
- Target = 1 → Customer experienced payment difficulties.

The dataset is imbalanced, with more successful repayments than defaults.

## 12. Data Quality Report

### Objective

Generate a summary report showing the quality of each column, including data type, missing values, missing percentage, and unique values.

```python
data_quality = pd.DataFrame({
    "Data Type": df.dtypes,
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": round((df.isnull().sum() / len(df)) * 100, 2),
    "Unique Values": df.nunique()
})

data_quality.head(20)
```

```python
data_quality.sort_values(
    by="Missing Percentage",
    ascending=False
).head(20)
```

### Observation

- Several housing-related variables contain a significant proportion of missing values.
- Financial and demographic variables are comparatively more complete.
- Missing values will require appropriate treatment before modeling and dashboard development.

## 13. Memory Usage
